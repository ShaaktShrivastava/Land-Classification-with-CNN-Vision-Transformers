# Memory-Based vs Generator-Based Image Loading


```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
import urllib.request
import tarfile

url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/4Z1fwRR295-1O3PMQBH6Dg/images-dataSAT.tar"
tar_path = "images-dataSAT.tar"

if not os.path.exists("images_dataSAT"):
    print("Downloading dataset...")
    urllib.request.urlretrieve(url, tar_path)
    print("Extracting...")
    with tarfile.open(tar_path) as tar:
        tar.extractall(".")
    print("Done.")
else:
    print("Dataset already exists, skipping download.")
```

```python
# set up paths to the two class folders
extract_dir = "."

base_dir = os.path.join(extract_dir, 'images_dataSAT')
dir_non_agri = os.path.join(base_dir, 'class_0_non_agri')
dir_agri = os.path.join(base_dir, 'class_1_agri')
```

```python
# scandir is a one-time iterator, so we restart it here
non_agri = os.scandir(dir_non_agri)

non_agri_images = []
for file_name in non_agri:
    if os.path.isfile(file_name):
        image_name = file_name.name
        image_data = plt.imread(os.path.join(dir_non_agri, image_name))
        non_agri_images.append(image_data)

non_agri_images = np.array(non_agri_images)
```
